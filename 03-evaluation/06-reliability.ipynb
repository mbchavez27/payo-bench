Analysis plan §9. R1 (annotation) is in notebook 01. R4's advice strategy-change part was dropped in v1.2, so advice variance is reported as reply-length spread only.

```python
# Shared setup: every path comes from payo_eval.paths, so PAYO_* variables can point the notebook at synthetic data
import warnings

import matplotlib.pyplot as plt
import pandas as pd

from payo_eval import analyses as A
from payo_eval import load
from payo_eval.paths import get_paths, load_config

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 200)
```

```python
# Settings from 03-evaluation/config.toml; outputs go to results/06-reliability/ and figures/
PATHS = get_paths()
CFG = load_config(PATHS)
SEED = CFG["seed"]
B = CFG["analysis"]["bootstrap_b"]
RESULTS = PATHS.results / "06-reliability"
TESTS = PATHS.results / "tests"
FIGURES = PATHS.figures
for folder in (RESULTS, TESTS, FIGURES):
    folder.mkdir(parents=True, exist_ok=True)


def save(table: pd.DataFrame, stem: str) -> pd.DataFrame:
    """Write a result table as CSV and return it for display."""
    table.to_csv(RESULTS / f"{stem}.csv", index=False)
    return table
```

```python
ratings = load.ratings(PATHS)
best = load.bestfit(PATHS)
```

```python
save(A.rating_reliability(ratings, best, CFG["analysis"]["alpha_tentative"]), "r2_r3")
```

```python
save(A.generation_variance(load.variance(PATHS), load.predictions(PATHS)), "r4_variance")
```
