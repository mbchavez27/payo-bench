# 07 · Multiple comparisons (Holm, §10)

Collects every confirmatory test written by notebooks 02-05 and applies Holm within each family. Run this last.

```python
# Shared setup: every path comes from payo_eval.paths, so PAYO_* variables can point the notebook at synthetic data
import warnings

import matplotlib.pyplot as plt
import pandas as pd

from payo_eval import analyses as A
from payo_eval import load
from payo_eval.paths import get_paths, load_config

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 200)
```

```python
# Settings from 03-evaluation/config.toml; outputs go to results/07-multiple-comparisons/ and figures/
PATHS = get_paths()
CFG = load_config(PATHS)
SEED = CFG["seed"]
B = CFG["analysis"]["bootstrap_b"]
RESULTS = PATHS.results / "07-multiple-comparisons"
TESTS = PATHS.results / "tests"
FIGURES = PATHS.figures
for folder in (RESULTS, TESTS, FIGURES):
    folder.mkdir(parents=True, exist_ok=True)


def save(table: pd.DataFrame, stem: str) -> pd.DataFrame:
    """Write a result table as CSV and return it for display."""
    table.to_csv(RESULTS / f"{stem}.csv", index=False)
    return table
```

```python
# Every test file from the other notebooks
files = sorted(TESTS.glob("F*.csv"))
print([f.name for f in files])
tests = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
```

```python
save(A.apply_holm(tests), "holm")
```
