# 05 · Linking question (H4a, H4b)

Analysis plan §6. H4a is observational (an association, with post-level controls and the within-post contrast). H4b is the interventional oracle test, read with the pre-specified rule.

```python
# Shared setup: every path comes from payo_eval.paths, so PAYO_* variables can point the notebook at synthetic data
import warnings

import matplotlib.pyplot as plt
import pandas as pd

from payo_eval import analyses as A
from payo_eval import load
from payo_eval.paths import get_paths, load_config

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 200)
```

```python
# Settings from 03-evaluation/config.toml; outputs go to results/05-linking/ and figures/
PATHS = get_paths()
CFG = load_config(PATHS)
SEED = CFG["seed"]
B = CFG["analysis"]["bootstrap_b"]
RESULTS = PATHS.results / "05-linking"
TESTS = PATHS.results / "tests"
FIGURES = PATHS.figures
for folder in (RESULTS, TESTS, FIGURES):
    folder.mkdir(parents=True, exist_ok=True)


def save(table: pd.DataFrame, stem: str) -> pd.DataFrame:
    """Write a result table as CSV and return it for display."""
    table.to_csv(RESULTS / f"{stem}.csv", index=False)
    return table
```

```python
ratings = load.ratings(PATHS)
```

## H4a: recognition → advice

```python
tests_a, model_a, within = A.recognition_to_advice(ratings, CFG["analysis"]["min_within_post"])
tests_a.to_csv(TESTS / "F4_h4a.csv", index=False)
save(model_a, "h4a_model")
```

```python
save(within, "h4a_within_post")
```

## H4b: oracle vs. P0

```python
tests_b, paired, reading = A.oracle_effect(ratings, B, SEED, CFG["analysis"]["margin_points"])
tests_b.to_csv(TESTS / "F4_h4b.csv", index=False)
save(paired, "h4b_paired")
```

```python
save(reading, "h4b_interpretation")
```

## Prompt-length confound

```python
save(A.prompt_tokens(load.replies(PATHS)), "prompt_tokens")
```
