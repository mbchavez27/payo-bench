# 04 · RQ2: strategy drift (H2b)

Analysis plan §4.2. Drift = P(autonomy | model) − P(autonomy | community) on the same posts, from the primary rater's strategy codes. Subgroups are descriptive only.

```python
# Shared setup: every path comes from payo_eval.paths, so PAYO_* variables can point the notebook at synthetic data
import warnings

import matplotlib.pyplot as plt
import pandas as pd

from payo_eval import analyses as A
from payo_eval import load
from payo_eval.paths import get_paths, load_config

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 200)
```

```python
# Settings from 03-evaluation/config.toml; outputs go to results/04-strategy-drift/ and figures/
PATHS = get_paths()
CFG = load_config(PATHS)
SEED = CFG["seed"]
B = CFG["analysis"]["bootstrap_b"]
RESULTS = PATHS.results / "04-strategy-drift"
TESTS = PATHS.results / "tests"
FIGURES = PATHS.figures
for folder in (RESULTS, TESTS, FIGURES):
    folder.mkdir(parents=True, exist_ok=True)


def save(table: pd.DataFrame, stem: str) -> pd.DataFrame:
    """Write a result table as CSV and return it for display."""
    table.to_csv(RESULTS / f"{stem}.csv", index=False)
    return table
```

```python
ratings = load.ratings(PATHS)
```

```python
tests, drift, groups = A.drift(ratings, B, SEED, CFG["analysis"]["min_cell"])
tests.to_csv(TESTS / "F2_h2b.csv", index=False)
save(drift, "drift")
```

## By domain and cultural concept (descriptive; cells under the minimum are flagged)

```python
save(groups, "drift_subgroups")
```

```python
# Strategy mix per source
mix = A.p0_level(ratings).query("slot == 1").groupby("source")["strategy"].value_counts(normalize=True).unstack()
ax = mix.plot.bar(stacked=True, figsize=(7, 4), rot=0)
ax.set_ylabel("share of replies")
ax.figure.tight_layout()
ax.figure.savefig(FIGURES / "04_strategy_mix.png", dpi=150)
save(mix.reset_index(), "strategy_mix")
```
